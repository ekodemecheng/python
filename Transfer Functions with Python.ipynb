# 2) SciPy.signal

```python
# SciPy.signal

import numpy as np                # library numpy
import scipy.signal as signal     # library SciPy.signal
import matplotlib.pyplot as plt   # plotting

# Define Transfer Function
num = np.array([2])
den = np.array([3 , 1])
H = signal.TransferFunction(num , den)
print ('H(s) =', H)

# Step Response
t, y = signal.step(H)

# Plotting
plt.plot(t, y)
plt.title("Step Response")
plt.xlabel("t")
plt.ylabel("y")
plt.grid()
plt.show()
```

# Time Constant T

```python
# import library
import numpy as np
import control
import matplotlib.pyplot as plt

K = 10
Tarray = [5, 10, 30, 50]
start = 0
stop = 200
step = 0.1
t = np.arange(start, stop, step)

for T in Tarray:
    
    #Create Transfer Function
    num = np.array ([K])
    den = np.array ([T , 1])
    H = control.tf(num , den)
    
    print ('H(s) =', H)

    # Step Response
    t, y = control.step_response(H, t)

    # Plot
    plt.plot(t, y)
    
plt.title("Step Response for different T")
plt.xlabel("t")
plt.ylabel("y")
plt.legend(Tarray)
plt.grid()
plt.show()

# Conclusion
# Larger 𝑇 → Slower System
# Smaller 𝑇 → Faster System
```

# Time Delay & Pade Approx

```python
# Import library
import numpy as np
import matplotlib.pyplot as plt
import control

# Time Delay
Tau = 2

# Approximation Order
N = 5

[num_pade,den_pade] = control.pade(Tau,N)
Hpade = control.tf(num_pade,den_pade)
print ('Hpade(s) =', Hpade)

start = 0
stop = 10
step = 0.1

t = np.arange(start, stop, step)

t, y = control.step_response(Hpade, t)

# Plot
plt.plot(t,y)
title = "Pade Aprroximation, N=" + str(N)
plt.title(title)
plt.grid()
```

```python
# import library
import numpy as np
import matplotlib.pyplot as plt
import control

# Time Delay
Tau = 2

N = 5

start = 0
stop = 10
step = 0.1
t = np.arange(start, stop, step)

N = [1, 2, 3, 5, 10]

for n in N:
    [num_pade,den_pade] = control.pade(Tau,n)
    Hpade = control.tf(num_pade,den_pade)
    print ('Hpade(s) =', Hpade)
    
    t, y = control.step_response(Hpade, t)
    
    # Plot
    plt.plot(t,y)
    
plt.title("Pade Aprroximations")
plt.legend(N)
plt.grid()
```

# First Order with Time Delay

```python
# import library
import numpy as np
import matplotlib.pyplot as plt
import control

K = 3
T = 4

num = np.array ([K])
den = np.array ([T , 1])
H1 = control.tf(num , den)
print ('H1(s) =', H1)

Tau = 2
N = 5 # Order of the Approximation
[num_pade,den_pade] = control.pade(Tau,N)
Hpade = control.tf(num_pade,den_pade)
print ('Hpade(s) =', Hpade)

H = control.series(H1, Hpade)
print ('H(s) =', H)

t, y = control.step_response(H)

plt.plot(t,y)
plt.title("H(s)")
plt.grid()
```

# 6) Poles and Zeros

```python
# Mode 1: Poles & Zeros

# import library
import numpy as np
import control

# define transfer function
num = np.array([1, 1])
den = np.array([3 , 1])
H1 = control.tf(num , den)

num = np.array([2, 1])
den = np.array([4 , 1])
H2 = control.tf(num , den)

H = control.series(H1,H2)
print ('H(s) =', H)

# Poles & zeros
z = control.zero(H)
print ('z =', z)
p = control.pole(H)
print ('p =', p)
control.pzmap(H)
```

```python
# Mode 2: Alternative Poles & Zeros

# import library
import numpy as np
import control

# define transfer function
num = np.array([2, 3, 1])
den = np.array([12 , 7, 1])
H = control.tf(num , den)

print ('H(s) =', H)

# Poles & zeros
z = control.zero(H)
print ('z =', z)
p = control.pole(H)
print ('p =', p)
control.pzmap(H)
```

```python
# Mode 3: Alternative Poles & Zeros, use np.convolve() function

# import library
import numpy as np
import control

# define transfer function
num1 = np.array([1, 1])
num2 = np.array([2, 1])
num = np.convolve(num1, num2)

den1 = np.array([3 , 1])
den2 = np.array([4 , 1])
den = np.convolve(den1, den2)

H = control.tf(num , den)

print ('H(s) =', H)

# Poles & zeros
z = control.zero(H)
print ('z =', z)
p = control.pole(H)
print ('p =', p)
control.pzmap(H)
```

# Poles and Stability of the System

```python
# Asymptotically Stable System

# import library
import control
import numpy as np
import matplotlib.pyplot as plt

# Define Transfer Function
num = np.array([2])
den = np.array([3 , 1])
H = control.tf(num , den)
print ('H(s) =', H)

# Poles
p = control.pole(H)
print ('p =', p)

# Step Response
t, y = control.step_response(H)

plt.plot(t,y)
plt.title("Step Response")
plt.grid()
```

```python
# Pole Zero Map: Asymptotically Stable System

control.pzmap(H)
```

```python
# Marginally Stable System

# import library
import control
import numpy as np
import matplotlib.pyplot as plt

# Define Transfer Function
num = np.array([3])
den = np.array([1, 0, 1])
H = control.tf(num , den)
print ('H(s) =', H)

# Poles
p = control.pole(H)
print ('p =', p)

# Step Response
tstart = 0; tstop = 20; tstep = 0.1
t = np.arange(tstart, tstop, tstep)
t, y = control.step_response(H, t)
plt.plot(t,y)
plt.title("Step Response")
plt.grid()
```

```python
# Pole Zero Map: Marginally Stable System

control.pzmap(H)
```
